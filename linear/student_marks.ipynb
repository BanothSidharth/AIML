```python
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
marks_df=pd.read_csv('student_scores.csv')
```

```python
fig = px.scatter(marks_df,x='Hours',y='Scores',title='Study Hours vs Scores',)
```

```python
inputs= marks_df[['Hours']]
targets = marks_df.Scores
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
model = LinearRegression()
```

```python
model.fit(inputs,targets)
```

```python
w=model.coef_
b = model.intercept_
print(w,b)
line = px.line(
    x=marks_df['Hours'],
    y=model.predict(marks_df[['Hours']]),color_discrete_sequence=['red']
)
fig.add_traces(line.data)
```

```python
y_preds= model.predict(marks_df[['Hours']])
error = targets-y_preds
ss_tot = ((targets-targets.mean())**2).sum()
mae = np.abs(error).mean()
rmse = np.sqrt(np.mean(error**2))
r_square = 1-(error**2).sum()/ss_tot
error,mae,rmse,r_square
```

```python
marks_df['Predicted Scores'] = model.predict(marks_df[['Hours']])
marks_df
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    inputs,
    targets,
    test_size=0.2,
    random_state=42
)
model.fit(X_train,y_train)
y_preds=model.predict(X_test)
print(X_test['Hours'])
y_preds
```

```python
import plotly.express as px

fig = px.scatter(
    x=X_test['Hours'],
    y=y_test,
    labels={'x': 'Study Hours', 'y': 'Scores'},
    title='Actual vs Predicted Student Scores'
)

fig.add_scatter(
    x=X_test['Hours'],
    y=y_preds,
    mode='markers',
    name='Predicted Scores',
    marker=dict(color='red', symbol='x', size=10)
)

fig.update_traces(
    selector=dict(name=''),
    name='Actual Scores',
    marker=dict(color='blue', size=10)
)

fig.show()
```
